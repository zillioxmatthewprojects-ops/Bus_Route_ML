Load Data and check info for types, total values, and miss match number

```python
#Created by Matt Z

#Starting Line up
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns


#import error bus data
df = pd.read_csv(r"/content/Bus Depot_data.csv")
print("Jumanji 1: Bus Data Loaded")


df.info()
```

Main error checking issues to identify duplicaes, column header issues and names, missing values, categorical unique values, and unique TripIDs.


```python
#1 check for duplicates
duplicate_count = df.duplicated().sum()
print(f"Jumanji 1: Duplicates:{duplicate_count}\n")

#2 check column header leading and training spacing /naming - capital or lower case
print("Jumanji 2: Column Headers")
print(df.columns.tolist())
print()

#3 check for missing values
missing_summary = df.isnull().sum()
print(f"Jumanji 3: Missing Values:{missing_summary}\n")

#4 Check min max values
Range_Check = df.describe()
print(f"Jumanji 4: Range Check:\n{Range_Check}\n")


#5 Determine variables, sample values, and other data issues
print("Jumanji 5: Categorical Unique Values")
for Z in df.columns:
    print(f"{Z}: {df[Z].dropna().unique()[:20]}\n")


#6 determine number of unique IDs
print("Jumanji 6: Unique TripID Values")
unique_ID = df['TripID'].nunique()
Total_Rows = len(df)

print(f"'Unique IDs:'{unique_ID}")
print(f"'Total rows:' {Total_Rows}")


# Unique minus total: negative = fewer unique IDs than rows,
# positive = more unique IDs than rows, 0 = every row has its own ID
diff = unique_ID - Total_Rows

if diff < 0:
    print(f"Too FEW unique IDs: {abs(diff)} fewer than rows (duplicate or blank IDs)")
elif diff > 0:
    print(f"Too MANY unique IDs: {diff} more than rows")
else:
    print("OK: every row has its own unique ID")
```

Found Issues

RouteName Issues: Various Route spellings, abreviations, and spacing

Solution: standardize capitalization (Makes RT = Rt), remove trailing spaces, and create a list of issues then replace from list.

```python
#Run this and add to the lists and steps as new errors are uincovered

#remove leading and trailing spaces, replace RT- with 'Route '  and capitalize
print("Jumanji 1: Fix route naming")
df['RouteName'] = (
    df['RouteName']
    .str.strip()
    .str.capitalize()
    .str.replace('Rt ', 'Route ', regex=False)
    .str.replace(r'\s+', ' ', regex=True)

)
print(df['RouteName'].unique())
print("")

#Determine what issues to fix from previous output and add them to the list here
print("Jumanji 2: Create list of issues and replace values complete")
state_fix = {
    'Ny':'New York', #Previous Error
    'Tx': 'Texas',   #Previous Error
    'Ca': 'California',
    'Fl':'Flordia',
    'Calif.':'California'
    }

print(state_fix)

print("\nJumanji 3: Fix State with making text tile case then replace from list")
df['State'] = (df['State']
               .str.title()
               .replace(state_fix)
               )

print(df['State'].unique())

#determine number of unique IDs and count of ids
print("\nJumanji 4: Display")


```
